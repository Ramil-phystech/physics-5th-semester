```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import constants as const
from IPython.display import Image
def angle(H):
    var = list()
    for i in range(len(H)):
        var.append(H[i][0] + 0.57 + H[i][1] / 60 + H[i][2] / 3600)
    return var
def angle_fin(H, H_max):
    b = (Zero - H_max + 180) / 2
    return Zero - b - H + 180
def angle_max(H_max):
    return (Zero - H_max + 180) / 2
def lambda_angle(H, H_max):
    return (np.sin(H*np.pi/180) - np.sin(H_max*np.pi/180)) * 834
def R_func(lambda_H, i):
    return 10**7 / (lambda_H*(1/4 - 1/(i**2)))
```

```python
for i in range(len(H_α)):
    H_α[i] = angle_fin(H_α[i], H_max[i])
for i in range(len(H_β)):
    H_β[i] = angle_fin(H_β[i], H_max[i])
for i in range(len(H_γ)):
    H_γ[i] = angle_fin(H_γ[i], H_max[i])
for i in range(len(H_max)):
    H_max[i] = angle_max(H_max[i])
```

## Постоянная Ридберга

```python
R_α = list()
for i in range(len(lambda_α)):
    R_α.append(R_func(lambda_α[i], 3))
R_β = list()
for i in range(len(lambda_β)):
    R_β.append(R_func(lambda_β[i], 4))
R_γ = list()
for i in range(len(lambda_γ)):
    R_γ.append(R_func(lambda_γ[i], 5))
```
